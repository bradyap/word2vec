## Reproduce the selected checkpoint

- Read `science.txt` from NLTK’s `abc` corpus. Use the sentence splitter and tokenizer in `word2vec.py` at revision `afb6120`.
- Shuffle sentences with seed **42** and split them 70% train, 15% validation, and 15% test. Build the vocabulary from the training sentences and generate center–context pairs with window size **2**.
- Use `min_count=5`, **64** embedding dimensions, batch size **64**, fixed rate **0.707**, and model seed **42**. Train a fresh model through epoch **10** with four BLAS threads.
- The full sweep screened fixed rates and dimensions, compared batches and nearby rates, checked the strongest configurations on a second model seed, and compared vocabulary cutoffs at the chosen settings.
- The selected checkpoint was scored once on held-out test pairs. The comparison baseline uses add-one-smoothed training context frequencies.
